## Steps

### 1. Tiny Example By Hand

If points rise from lower left to upper right, two quantities have a positive relationship. Correlation summarizes that pattern, but it does not prove one variable directly causes the other.

Before running Python, make a prediction: should the answer describe a typical
object, the amount of spread, the strength of a pattern, or the uncertainty in
an observation?


```python
temperature = np.array([3200, 3800, 4500, 5200, 5800, 6500, 7600, 9000])
luminosity = np.array([0.03, 0.08, 0.25, 0.55, 1.0, 2.0, 6.5, 22.0])

correlation = np.corrcoef(temperature, luminosity)[0, 1]
print("Correlation:", correlation)

plt.figure(figsize=(7, 4))
plt.scatter(temperature, luminosity, s=90)
plt.title("Temperature And Luminosity In A Toy Star Sample")
plt.xlabel("Temperature (K)")
plt.ylabel("Luminosity (solar units)")
plt.show()
```
