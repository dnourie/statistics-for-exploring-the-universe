# 09. Signal Or Noise?

## Goal

Astronomy question: **Is a small dip in a star's brightness a possible exoplanet transit, or just noise?**

Statistics enters because real astronomical observations are incomplete, noisy,
variable, and uncertain. In this notebook, we will use a small astronomy example
to make the statistical idea feel useful before we let Python handle the arithmetic.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-whitegrid")
np.random.seed(42)
```

## Steps

### 1. Tiny Example By Hand

The null idea says: suppose there is no planet and the star is steady. Would the observed dip be unusual under that assumption?

Before running Python, make a prediction: should the answer describe a typical
object, the amount of spread, the strength of a pattern, or the uncertainty in
an observation?


```python
time = np.linspace(0, 6, 80)
baseline_noise = np.random.normal(0, 0.004, size=len(time))
transit = np.where((time > 2.6) & (time < 3.3), -0.018, 0.0)
brightness = 1.0 + baseline_noise + transit

outside_transit = brightness[(time < 2.4) | (time > 3.5)]
inside_transit = brightness[(time > 2.6) & (time < 3.3)]

dip = outside_transit.mean() - inside_transit.mean()
noise_level = outside_transit.std(ddof=1)
signal_to_noise = dip / noise_level

print("Estimated dip:", dip)
print("Outside-transit noise level:", noise_level)
print("Simple signal-to-noise ratio:", signal_to_noise)

plt.figure(figsize=(8, 4))
plt.plot(time, brightness, marker="o", linestyle="-")
plt.axvspan(2.6, 3.3, color="tab:orange", alpha=0.2, label="possible transit")
plt.title("Toy Exoplanet Transit Signal")
plt.xlabel("Time (hours)")
plt.ylabel("Relative brightness")
plt.legend()
plt.show()
```
