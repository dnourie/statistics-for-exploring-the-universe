## Steps

### 1. Tiny Example By Hand

If a faint source produces about 3 photons per short exposure, one exposure might record 1 photon and another might record 5. Randomness is part of the measurement.

Before running Python, make a prediction: should the answer describe a typical
object, the amount of spread, the strength of a pattern, or the uncertainty in
an observation?


```python
photon_counts = np.random.poisson(lam=3, size=30)

print("Counts:", photon_counts)
print("Average count:", photon_counts.mean())

plt.figure(figsize=(7, 4))
plt.bar(range(1, len(photon_counts) + 1), photon_counts)
plt.axhline(3, color="tab:red", label="expected count = 3")
plt.title("Photon Counts From A Faint Source")
plt.xlabel("Exposure number")
plt.ylabel("Detected photons")
plt.legend()
plt.show()
```
