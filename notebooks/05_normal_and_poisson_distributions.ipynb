# 05. Normal And Poisson Distributions

## Goal

Astronomy question: **Why do measurement errors often look bell-shaped, while photon counts look count-like and uneven?**

Statistics enters because real astronomical observations are incomplete, noisy,
variable, and uncertain. In this notebook, we will use a small astronomy example
to make the statistical idea feel useful before we let Python handle the arithmetic.


## Setup

Concepts in this lesson:

- normal distribution for many measurement errors
- Poisson distribution for counts
- choosing a model that matches the science

Keep the science question in view while you work: the statistic is not just a
number. It is a compact way of saying something about light, objects, or
measurements from the universe.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-whitegrid")
np.random.seed(42)
```

## Steps

### 1. Tiny Example By Hand

A brightness error can be above or below zero, so a bell-shaped normal model can make sense. A photon count cannot be negative, so a count model is better.

Before running Python, make a prediction: should the answer describe a typical
object, the amount of spread, the strength of a pattern, or the uncertainty in
an observation?


```python
normal_errors = np.random.normal(loc=0, scale=1, size=1000)
poisson_counts = np.random.poisson(lam=3, size=1000)

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].hist(normal_errors, bins=30, edgecolor="white")
axes[0].set_title("Normal-Like Measurement Errors")
axes[0].set_xlabel("Error in brightness units")
axes[0].set_ylabel("Frequency")

axes[1].hist(poisson_counts, bins=np.arange(-0.5, 12.5, 1), edgecolor="white")
axes[1].set_title("Poisson-Like Photon Counts")
axes[1].set_xlabel("Detected photons")
axes[1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```
