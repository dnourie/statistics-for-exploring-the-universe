```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-whitegrid")
np.random.seed(42)
```

## Steps

### 1. Tiny Example By Hand

A sample mean changes from sample to sample. That does not mean the method is broken; it means limited data creates uncertainty.

Before running Python, make a prediction: should the answer describe a typical
object, the amount of spread, the strength of a pattern, or the uncertainty in
an observation?


```python
population_masses = np.random.lognormal(mean=10.3, sigma=0.35, size=5000)
sample = np.random.choice(population_masses, size=40, replace=False)

sample_mean = sample.mean()
sample_standard_error = sample.std(ddof=1) / np.sqrt(len(sample))
interval = (sample_mean - 2 * sample_standard_error, sample_mean + 2 * sample_standard_error)

print("Sample mean:", sample_mean)
print("Approximate 95% interval:", interval)

plt.figure(figsize=(7, 4))
plt.hist(sample, bins=12, edgecolor="white")
plt.axvline(sample_mean, color="tab:red", label="sample mean")
plt.title("A Sample Of Simulated Galaxy Masses")
plt.xlabel("Mass proxy")
plt.ylabel("Number of galaxies")
plt.legend()
plt.show()
```
