## Goal

Astronomy question: **If we measure the same star many times, why do we not get exactly the same brightness every time?**

Statistics enters because real astronomical observations are incomplete, noisy,

to make the statistical idea feel useful before we let Python handle the arithmetic.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-whitegrid")
np.random.seed(42)
```

## Steps

### 1. Tiny Example By Hand

A telescope might measure one steady star as $1.01, 0.99, 1.03, 0.98$. The star did not necessarily change; the measurements wobble around a central value.

Before running Python, make a prediction: should the answer describe a typical
object, the amount of spread, the strength of a pattern, or the uncertainty in
an observation?


```python
measurements = np.array([1.01, 0.99, 1.03, 0.98, 1.00, 1.02, 0.97, 1.01])

print("Mean brightness:", measurements.mean())
print("Standard deviation:", measurements.std(ddof=1))

plt.figure(figsize=(7, 4))
plt.plot(measurements, marker="o")
plt.axhline(measurements.mean(), color="tab:red", label="mean")
plt.fill_between(
    range(len(measurements)),
    measurements.mean() - measurements.std(ddof=1),
    measurements.mean() + measurements.std(ddof=1),
    color="tab:red",
    alpha=0.15,
    label="about 1 standard deviation",
)
plt.title("Repeated Brightness Measurements Of One Star")
plt.xlabel("Observation number")
plt.ylabel("Relative brightness")
plt.legend()
plt.show()
```
