```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-whitegrid")
np.random.seed(42)
```

## Steps

### 1. Tiny Example By Hand

If planet detections are rare, a weak signal should not immediately convince us. Strong evidence can still overcome a skeptical prior.

Before running Python, make a prediction: should the answer describe a typical
object, the amount of spread, the strength of a pattern, or the uncertainty in
an observation?


```python
prior_planet = 0.05
likelihood_signal_if_planet = 0.90
likelihood_signal_if_no_planet = 0.08

unnormalized_planet = prior_planet * likelihood_signal_if_planet
unnormalized_no_planet = (1 - prior_planet) * likelihood_signal_if_no_planet
posterior_planet = unnormalized_planet / (unnormalized_planet + unnormalized_no_planet)

print("Prior probability of planet:", prior_planet)
print("Posterior probability after signal:", posterior_planet)

labels = ["Before evidence", "After evidence"]
values = [prior_planet, posterior_planet]

plt.figure(figsize=(6, 4))
plt.bar(labels, values)
plt.ylim(0, 1)
plt.title("Bayesian Update For A Possible Planet")
plt.ylabel("Probability")
plt.show()
```
