```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-whitegrid")
np.random.seed(42)
```

## Steps

### 1. Tiny Example By Hand

A list of brightnesses is hard to read one number at a time. A histogram turns the list into a visual summary of where values pile up.

Before running Python, make a prediction: should the answer describe a typical
object, the amount of spread, the strength of a pattern, or the uncertainty in
an observation?


```python
ordinary_stars = np.random.normal(loc=1.0, scale=0.12, size=90)
bright_giants = np.random.normal(loc=2.2, scale=0.25, size=10)
brightness = np.concatenate([ordinary_stars, bright_giants])

plt.figure(figsize=(7, 4))
plt.hist(brightness, bins=18, edgecolor="white")
plt.title("Distribution Of Relative Stellar Brightness")
plt.xlabel("Relative brightness")
plt.ylabel("Number of stars")
plt.show()

print("Mean:", brightness.mean())
print("Median:", np.median(brightness))
```
