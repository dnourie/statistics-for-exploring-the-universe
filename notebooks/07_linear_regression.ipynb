```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-whitegrid")
np.random.seed(42)
```

## Steps

### 1. Tiny Example By Hand

A line has the form $y = mx + b$. The slope $m$ says how much $y$ changes when $x$ increases by 1 unit. The intercept $b$ is the model prediction when $x = 0$.

Before running Python, make a prediction: should the answer describe a typical
object, the amount of spread, the strength of a pattern, or the uncertainty in
an observation?


```python
distance_pc = np.array([10, 20, 30, 40, 50, 60, 70])
redshift_proxy = np.array([0.003, 0.006, 0.010, 0.013, 0.016, 0.019, 0.023])

slope, intercept = np.polyfit(distance_pc, redshift_proxy, 1)
predicted = slope * distance_pc + intercept
residuals = redshift_proxy - predicted

print("Slope:", slope)
print("Intercept:", intercept)

plt.figure(figsize=(7, 4))
plt.scatter(distance_pc, redshift_proxy, label="observed")
plt.plot(distance_pc, predicted, color="tab:red", label="fit")
plt.title("A Toy Linear Relationship In Astronomy")
plt.xlabel("Distance (parsecs)")
plt.ylabel("Redshift proxy")
plt.legend()
plt.show()

display(pd.DataFrame({
    "distance_pc": distance_pc,
    "observed": redshift_proxy,
    "predicted": predicted,
    "residual": residuals,
}))
```
