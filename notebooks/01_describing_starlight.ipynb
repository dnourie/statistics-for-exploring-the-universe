# Describing Starlight

## Astronomy Question

**If our measurements are slightly different every time, what value best
represents the brightness of the star?**

Imagine we point a telescope at one steady star five times. We expect the star
to have one real brightness, but our measurements do not match perfectly. That
does not mean the star is changing every few seconds. It may simply mean our
instrument, atmosphere, detector, or observing conditions added tiny wiggles to
the numbers.

This lesson introduces statistics as a language for describing those wiggles.
We are not doing statistics for its own sake. We are trying to say something
careful about starlight.


## What You Will Learn

By the end, you will know how to describe a small set of repeated brightness
measurements using:

- mean
- median
- range
- spread
- variance, conceptually
- standard deviation, conceptually and computationally
- the effect of an outlier on the mean and median

For now, the goal is to become comfortable asking: **what does this number
tell me about the observation?**


## 1. A Tiny Dataset We Can Understand By Hand

Here are five repeated measurements of the same star's relative brightness.

Relative brightness means we are measuring the star compared with a reference
level. A value near `1.00` means the star is close to its usual measured
brightness.

| Observation | Relative brightness |
|---:|---:|
| 1 | 0.99 |
| 2 | 1.01 |
| 3 | 1.00 |
| 4 | 1.02 |
| 5 | 0.98 |

Before calculating anything, notice the astronomy story: every value is close
to `1.00`, but no two readings are guaranteed to be identical. We need a
reasonable single number to represent the star, and we also need a way to
describe how much the measurements vary.


```python
brightness = [0.99, 1.01, 1.00, 1.02, 0.98]
brightness
```

## 2. Mean: The Balance Point

The **mean** is the ordinary average.

For our five brightness measurements:

$$
\text{mean} = \frac{0.99 + 1.01 + 1.00 + 1.02 + 0.98}{5}
$$

Add the measurements:

$$
0.99 + 1.01 + 1.00 + 1.02 + 0.98 = 5.00
$$

Now divide by the number of measurements:

$$
\frac{5.00}{5} = 1.00
$$

So the mean brightness is **1.00 relative brightness units**.

Astronomy meaning: if we had to choose one number to summarize these repeated
measurements, the mean says the star's measured brightness is centered around
`1.00`.


```python
mean_brightness = sum(brightness) / len(brightness)
print("Mean brightness:", mean_brightness)
```

The Python result matches the hand calculation.

Astronomy meaning: the telescope readings balance around `1.00`. This is a
good first estimate of the star's brightness when the measurements are tightly
clustered and there are no suspicious values.


## 3. Median: The Middle Measurement

The **median** is the middle value after sorting the data.

Sort the brightness measurements:

$$
0.98,\ 0.99,\ 1.00,\ 1.01,\ 1.02
$$

The middle value is:

$$
1.00
$$

So the median brightness is **1.00 relative brightness units**.

Astronomy meaning: half the measurements are below `1.00` and half are above
`1.00`. The median is another way to say what a typical reading looks like.


```python
sorted_brightness = sorted(brightness)
middle_index = len(sorted_brightness) // 2
median_brightness = sorted_brightness[middle_index]

print("Sorted measurements:", sorted_brightness)
print("Median brightness:", median_brightness)
```

Here, the mean and median agree exactly.

Astronomy meaning: our measurements are nicely centered. When the mean and
median are close, that is often a sign that one strange measurement is not
dominating the summary.


## 4. Range: The Full Width Of The Measurements

The **range** is:

$$
\text{range} = \text{largest value} - \text{smallest value}
$$

For the star:

$$
1.02 - 0.98 = 0.04
$$

So the range is **0.04 relative brightness units**.

Astronomy meaning: across these five observations, the measured brightness
moved from `0.98` to `1.02`. The full observed wiggle is `0.04`, even though
the star's best representative value is near `1.00`.


```python
brightness_range = max(brightness) - min(brightness)
print("Smallest brightness:", min(brightness))
print("Largest brightness:", max(brightness))
print("Range:", brightness_range)
```

The range tells us the total span of the measurements.

Astronomy meaning: this telescope session measured a brightness band only
`0.04` units wide. The range is simple and useful, but it only uses the two
most extreme measurements.


## 5. Visualizing The Measurements

A plot helps us see the same story at a glance. Each point is one telescope
measurement. The horizontal line marks the mean.


```python
import matplotlib.pyplot as plt

observation_numbers = [1, 2, 3, 4, 5]

plt.figure(figsize=(8, 4.5))
plt.plot(observation_numbers, brightness, marker="o", linewidth=2)
plt.axhline(mean_brightness, color="tab:red", linestyle="--", label="mean = 1.00")
plt.title("Repeated Brightness Measurements Of One Star")
plt.xlabel("Observation number")
plt.ylabel("Relative brightness")
plt.ylim(0.96, 1.04)
plt.legend()
plt.show()
```

What to notice:

The measurements wobble slightly above and below the mean. In astronomy
language, the star looks steady at about `1.00` relative brightness units, and
the small up-and-down pattern is the measurement spread.


## 6. Spread: How Far The Measurements Wander

The **spread** tells us how much the measurements vary around a representative
value.

Two stars could have the same mean brightness but very different spread:

- Star A: measurements tightly packed near the mean
- Star B: measurements scattered farther from the mean

Astronomy meaning: spread helps us describe uncertainty in the observation. If
the spread is small, repeated measurements agree closely. If the spread is
large, we should be more cautious about saying exactly how bright the star is.


```python
deviations = [value - mean_brightness for value in brightness]

for observation, value, deviation in zip(observation_numbers, brightness, deviations):
    print(f"Observation {observation}: brightness = {value:.2f}, deviation from mean = {deviation:+.2f}")
```

Each deviation says how far one measurement is from the mean.

Astronomy meaning: the first observation, `0.99`, is `0.01` below the mean. The
fourth observation, `1.02`, is `0.02` above the mean. These deviations are the
raw ingredients for describing spread.


## 7. Visual Intuition: Same Mean, Different Spread

Now compare two imaginary observing runs. Both stars have the same mean
brightness, but one set of measurements is packed tightly around the mean and
the other wanders farther away.

This is the idea behind spread: two datasets can agree on the center but differ
in how stable the measurements look.


```python
tight_measurements = [0.99, 1.00, 1.01, 1.00, 1.00]
wide_measurements = [0.94, 1.06, 0.97, 1.08, 0.95]

tight_mean = sum(tight_measurements) / len(tight_measurements)
wide_mean = sum(wide_measurements) / len(wide_measurements)

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5), sharey=True)

axes[0].scatter(observation_numbers, tight_measurements, s=90)
axes[0].axhline(tight_mean, color="tab:red", linestyle="--", label=f"mean = {tight_mean:.2f}")
axes[0].set_title("Tightly Grouped Measurements")
axes[0].set_xlabel("Observation number")
axes[0].set_ylabel("Relative brightness")
axes[0].legend()

axes[1].scatter(observation_numbers, wide_measurements, s=90)
axes[1].axhline(wide_mean, color="tab:red", linestyle="--", label=f"mean = {wide_mean:.2f}")
axes[1].set_title("More Spread Out Measurements")
axes[1].set_xlabel("Observation number")
axes[1].legend()

plt.ylim(0.90, 1.10)
plt.tight_layout()
plt.show()

print("Tight mean:", tight_mean)
print("Wide mean:", wide_mean)
```

What to notice:

Both observing runs have a mean near `1.00`, but they do not tell the same
story. The tightly grouped measurements make the brightness estimate feel more
stable. The wider measurements suggest the observing run had more scatter.

Astronomy meaning: a representative brightness value is only part of the
description. We also need to know how much the measurements move around that
value.


## 8. Variance: Squared Spread

The **variance** is based on squared deviations from the mean.

Why square the deviations?

- It prevents positive and negative deviations from canceling each other out.
- It gives larger wiggles more weight than tiny wiggles.

For this first lesson, think of variance as **the average squared distance from
the mean**. Its units are squared brightness units, so it is useful for
calculation, but less intuitive to read directly.


```python
squared_deviations = [deviation ** 2 for deviation in deviations]
variance = sum(squared_deviations) / len(squared_deviations)

for observation, deviation, squared in zip(observation_numbers, deviations, squared_deviations):
    print(f"Observation {observation}: deviation = {deviation:+.2f}, squared deviation = {squared:.4f}")

print("Variance:", variance)
```

Astronomy meaning: the variance is very small because all five brightness
measurements are close to the mean. But because variance is measured in
**squared** brightness units, it is not the easiest number to interpret by
itself.

That is why we usually move one more step to standard deviation.


## 9. Standard Deviation: Typical Distance From The Mean

The **standard deviation** is the square root of the variance.

Conceptually, it answers:

**How far from the mean is a typical measurement?**

Because we take the square root, standard deviation returns to the original
units: relative brightness units.


```python
standard_deviation = variance ** 0.5
print("Standard deviation:", standard_deviation)
```

The standard deviation is about `0.014` relative brightness units.

Astronomy meaning: a typical measurement in this observing run is about `0.014`
brightness units away from the mean brightness of `1.00`. This gives us a
compact way to say, "The star is around `1.00`, with small measurement-to-
measurement wiggles."


## 10. Visualizing Spread Around The Mean

This plot shows the mean and a band one standard deviation above and below the
mean.


```python
upper_band = mean_brightness + standard_deviation
lower_band = mean_brightness - standard_deviation

plt.figure(figsize=(8, 4.5))
plt.scatter(observation_numbers, brightness, s=90, label="measurements")
plt.axhline(mean_brightness, color="tab:red", linestyle="--", label="mean")
plt.fill_between(
    observation_numbers,
    lower_band,
    upper_band,
    color="tab:red",
    alpha=0.15,
    label="mean ± 1 standard deviation",
)
plt.title("Brightness Measurements With A Spread Band")
plt.xlabel("Observation number")
plt.ylabel("Relative brightness")
plt.ylim(0.96, 1.04)
plt.legend()
plt.show()
```

What to notice:

Most measurements sit close to the shaded band. The band is not magic; it is a
visual guide for the typical size of the measurement wiggles.

Astronomy meaning: the star's repeated readings are consistent with a steady
brightness near `1.00`, measured with small observational spread.


## 11. Reproducing The Calculations With Python Tools

Python has built-in tools and scientific libraries that can calculate these
summaries quickly. We will use them here only after understanding the ideas by
hand.


```python
import numpy as np

brightness_array = np.array(brightness)

print("Mean:", np.mean(brightness_array))
print("Median:", np.median(brightness_array))
print("Range:", np.max(brightness_array) - np.min(brightness_array))
print("Variance:", np.var(brightness_array))
print("Standard deviation:", np.std(brightness_array))
```

These Python calculations summarize the same observing run.

Astronomy meaning:

- Mean and median tell us the representative brightness.
- Range tells us the total observed brightness span.
- Variance and standard deviation tell us how much the measurements spread out
  around the representative brightness.

For a beginner, standard deviation is usually the most readable spread number
because it stays in the same units as the brightness measurements.


## 12. What Happens When There Is An Outlier?

Now imagine one observation is unusually high:

$$
0.99,\ 1.01,\ 1.00,\ 1.02,\ 1.18
$$

Maybe a cosmic ray hit the detector. Maybe a cloud briefly affected the
reference star. Maybe the data reduction had a problem. Or maybe something real
happened. At this stage, we only know that `1.18` does not look like the other
four readings.

An **outlier** is a value that is far away from the rest of the data.


```python
brightness_with_outlier = [0.99, 1.01, 1.00, 1.02, 1.18]

mean_with_outlier = np.mean(brightness_with_outlier)
median_with_outlier = np.median(brightness_with_outlier)

print("Original mean:", np.mean(brightness_array))
print("Mean with outlier:", mean_with_outlier)
print()
print("Original median:", np.median(brightness_array))
print("Median with outlier:", median_with_outlier)
```

The outlier changes the mean more than it changes the median.

Astronomy meaning: the mean is pulled toward the unusually bright measurement.
The median stays close to the central cluster of observations. This is why
astronomers often compare mean and median when checking whether a summary is
being distorted by one unusual data point.


## 13. Visualizing The Outlier Effect

The next plot compares the original measurements with the outlier version.


```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5), sharey=True)

axes[0].scatter(observation_numbers, brightness, s=90)
axes[0].axhline(np.mean(brightness), color="tab:red", linestyle="--", label="mean")
axes[0].axhline(np.median(brightness), color="tab:green", linestyle=":", label="median")
axes[0].set_title("Original Measurements")
axes[0].set_xlabel("Observation number")
axes[0].set_ylabel("Relative brightness")
axes[0].legend()

axes[1].scatter(observation_numbers, brightness_with_outlier, s=90)
axes[1].axhline(mean_with_outlier, color="tab:red", linestyle="--", label="mean")
axes[1].axhline(median_with_outlier, color="tab:green", linestyle=":", label="median")
axes[1].set_title("With One High Outlier")
axes[1].set_xlabel("Observation number")
axes[1].legend()

plt.ylim(0.96, 1.22)
plt.tight_layout()
plt.show()
```

What to notice:

In the original data, the mean and median sit together. With the outlier, the
mean moves upward more noticeably, while the median remains near the middle of
the ordinary measurements.

Astronomy meaning: if a single measurement is suspicious, the median may better
describe the typical brightness. But we should not automatically delete the
outlier. In astronomy, unusual points can be errors, but they can also be
discoveries.


## 14. Lesson Summary

For repeated measurements of one star:

- The **mean** is the balance-point brightness.
- The **median** is the middle measured brightness.
- The **range** is the full width from dimmest measurement to brightest
  measurement.
- **Spread** describes how much the measurements vary.
- **Variance** is based on squared distances from the mean.
- **Standard deviation** is the typical distance from the mean, back in the
  original brightness units.
- An **outlier** can pull the mean more strongly than the median.

The astronomy interpretation is the heart of the lesson: we are using these
numbers to describe what the telescope saw and how much trust to place in one
representative brightness value.
